```python
#What is exception handling in Python?
```

```python
try:
    print(10 / 0)
except ZeroDivisionError:
    print("Cannot divide by zero")
```

```python
#What is the purpose of the finally block in exception handling?
```

```python
try:
    print(10 / 0)
except ZeroDivisionError:
    print("Error occurred")
finally:
    print("Finally block executed")
```
